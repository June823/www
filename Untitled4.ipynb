```python
import spacy

# Load small English model
nlp = spacy.load("en_core_web_sm")
```

```python
reviews = [
    "I love the Apple iPhone 14 Pro Max. The camera quality is amazing!",
    "This Samsung TV stopped working after two weeks. Terrible product.",
    "The Sony headphones are lightweight and have great sound.",
    "I bought the HP laptop and it's very slow. Not happy.",
]
```

```python
for review in reviews:
    doc = nlp(review)
    print("Review:", review)
    print("Entities found:")

    for ent in doc.ents:
        print(f" - {ent.text} ({ent.label_})")

    print("\n")
```

```python
positive_words = ["love", "amazing", "great", "excellent", "happy"]
negative_words = ["terrible", "slow", "bad", "worst", "not happy", "stopped"]

def get_sentiment(text):
    text_lower = text.lower()
    score = 0

    for word in positive_words:
        if word in text_lower:
            score += 1

    for word in negative_words:
        if word in text_lower:
            score -= 1

    if score > 0:
        return "Positive"
    elif score < 0:
        return "Negative"
    else:
        return "Neutral"
```

```python
for review in reviews:
    sentiment = get_sentiment(review)
    print(f"Review: {review}")
    print(f"Sentiment: {sentiment}\n")
```
